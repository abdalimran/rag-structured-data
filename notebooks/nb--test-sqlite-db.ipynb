```python
import sqlite3

# Connect to an existing database or create a new one if it doesn't exist
conn = sqlite3.connect("data/databases/chinook_sample.sqlite3")
```

```python
if conn:
    print("Database connection is active.")
    print(f"Connection object: {conn}")
    print(f"Database version: {conn.execute('SELECT sqlite_version()').fetchone()[0]}")
else:
    print("Database connection is not active.")
```

```python
# Get avaliable tables
cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()

print("Available tables:")
for table in tables:
    print(table[0])
```

```python
# Create a cursor object to execute SQL commands
cursor = conn.cursor()

# Query data
cursor.execute("SELECT * FROM Album")
rows = cursor.fetchall()
for row in rows:
    print(row)

# Close the connection when done
conn.close()
```
